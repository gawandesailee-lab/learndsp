## 1.2 Introduction<a id='1.2_Introduction'></a>

Ultimate Inc. operates in the neighboring cities of Metropolis and Gotham. The two cities are separated by a two way toll bridge and hence drivers do not like crossing the bridge for additional out of pocket incurred toll expenses. They operate in their own inidvidual cities of Metropolis or Gotham. 

Also, one point worth noting is that both cities operate in a circadian rhythm especially during weekdays. ex. iF Metropolis is active during daytime, Gotham is quiet as far as user activity is concerned and vice versa. Both cities have a more steady spread out demand across day/night on weekends.

Ultimate Inc. is wondering if they could take advantage of these facts by paying for the driver tolls for cross city rides thus encouraging drivers to operate intercity and also both weedays and weeknights since, one city is more active at nights and the other one in the mornings.

They would like to test this plan to ensure their assumptions of inter city toll being the only key factor for increased user demand and driver services to increase their Return on Investment is correct, sufficient enough to boost their current profits and increase both user and driver usage and satisfaction of their services.

## 1.3 Key Measure of Success<a id='1.3_Key_Measure_of_Success'></a>

Since, the key question here is as follows

#### Does complete reimbursement of inter city toll for all drivers boost user and driver demand of their services through out the 24 hour cycle regardless of the city the driver is currently operating in?

To answer this question we need to measure the following

#### % of Drivers operating in both cities post the toll reimbursement strategy implementation.

### 1.3.1 What is our KPI?<a id='1.3.1_What_is_our_KPI?'></a>

So, our Primary KPI would be as follows

#### Driver Cross City Participation Rate = % of Drivers operating in both cities / Total number of active drivers 

during a given time period (example weekly).

#### Secondary KPIs that would provide a more holistic view would be as follows

1. Driver Utilization : Drivers providing active service / hour
2. Trip Completion Rate : Active Driver completed trips during the time frame being tested
4. Avg. Driver Earnings after toll reimbursements
5. Surge in User Wait Times
6. Cost per incremental inter city Driver
7. % Cancellation Increase

### 1.3.2 Logic for Choosing the KPI?<a id='1.3.2_Logic_for_Choosing_the_KPI?'></a>

The logic for choosing our Primary KPI is to check if removing the toll payment roadblock is motivating enough for drivers to take intercity trips or not.

If not, are there some additional patterns emerging regarding other roadblocks for drivers to provide inter city rides.

## 1.4 Proposed Experiment<a id='1.4_Proposed_Experiment'></a>

The proposed experiment is as follows

### 1.4.1 The Experiment<a id='1.4.1_The_Experiment?'></a>

Randomly assign roughly equal number of active drivers across both cities to Test group and Control Groups (A/B Testing). Unit of randomization needs to be at driver level to avoid contamination, ensure causal inference regarding adoption behavior.

Test Group will be reimbursed all tolls.

Control Group will not be reimbursed any tolls.

### 1.4.2 Implementation Proposal<a id='1.4.2_Implementation_Proposal'></a>

1. Implement this experiment for 4 to 6 weeks so multiple weekday/weekend cycles get tested.  
2. Both groups should have similar characteristics, ex. both groups should have roughly equal number of each city active drivers, similar trip volumes, simlar operating time frames.  
3. This will be introduced as a blind trial. i.e. Drivers will not be told upfront if they will be reimbursed (i.e drivers should not know which group they belong to). However given the nature of the experiment, they will know as soon as they get reimbursed regarding which group they belong to.  
Reimbursement should be immediately done on toll crossing to motivate drivers to do inter city trips.

### 1.4.3 Statistical Tests <a id='1.4.3_Statistical_Tests'></a>

1.) Two proportion Z test would be our key statistical test for this experiment as we would be testing the multi city participation impact for both groups.

#### Null Hypothesis : 
Test/Control group did not show significant difference in cross city participation despite toll reimbursement motivation.  
#### Alternative Hypothesis : 
Test group showed a significant difference (increase) in cross city participation compared to the control group based on toll reimbursement.<p>  

2.) We will also build and test a logistic regression model given the nature of our experiment with repeated weekly observations.\
Driver-Week will be categorized into yes/no groups (binary outcome) for cross city driving as the dependent variable. Our primary independent variable will be whether the driver got reimbursed or not.Driver home city, time of day active, day of week active, trip volume percentile will be some of the other features that will be potentially included to improve model precision.

### 1.4.4 Results Interpretation <a id='1.4.4_Results_Interpretation'></a>

We would know that the experiment is successful in increasing the driver cross city participation if  
1. the 2 proportion z test shows significant increase in cross city participation for test (reimbursed tolls) group vs. the control group (no toll reimbursements without any major negative effects.
2. Logistic Regression model key prediction variable (reimbursed or not) is statistically significant and positive indicating increased inter city driver participation.

## 1.5 Key Recommendations<a id='1.5_Key_Recommendations'></a>

If the experiment is successful based on teh results interpretation above and there is a significant increase in cross city driver participation without any major effects on baseline company costs and ROI, then the recommendation would be to do a complete rollout for both cities.

However, if the results are mixed ex. one city drivers participating way more for cross city trips or not much difference is observed across the board for increased cross city driver trips, then partial or controlled rollout is recommended along with inspection of the business assumption that toll fee is the roadblocker for cross city trips.

## 1.6 Caveats<a id='1.6_Caveats'></a>

1. Cost Vs. Benefits Analysis should be the key driver for rolling out this cross city plan.
2. Drivers will try out this initiative by doing cross city trips initially to understand if it benefits them. If there is not significant monetary benefit with way more additional time on road then the plan will fizzle out in a few weeks or even a few months.
3. Drivers from both groups may interact and influence each other negatively for cross city trips.

## 1.7 Summary<a id='1.7_Summary'></a>

#### 1.	 Goal: 
Test whether toll reimbursement increases inter-city driver activity between Gotham and Metropolis
 
#### 2.	Metric: 
% of drivers serving both cities
  
#### 3.	Experiment Design: 
i)	Randomized Controlled Trial at Driver Level (test group  gets reimbursement, control does not get reimbursed)
ii)	This will ensure causal inference

#### 4.	Analysis Approach: 
i)	Two-proportion z-test (one-sided) to detect increase in inter city driver participation
ii)	Logistic regression model at driver week level with intercity participation (binary outcome) as the dependent variable, reimbursed or not as the key independent variable and additional controls like home city, day and time-of-week to improve precision


#### 5.	Decision Rule: 
i)	Significant increase in intercity driver participation→ recommend rollout 
ii)	No effect → reconsider incentive design and look at other variables negatively impacting intercity driver participation


#### 6.	Key Caveats: 
i)	Cost vs. benefit tradeoff  
ii)	Short Term vs. Long Term Benefits  
iii)Drivers from one group influencing another group

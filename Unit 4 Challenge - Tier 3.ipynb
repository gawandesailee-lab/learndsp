**3. Modeling**

Consider creating a function that will calculate a ratio of house prices, comparing the price of a house in 2018 to the price in 1998.

Consider calling this function create_price_ratio.

You'd want this function to:
1. Take a filter of dfg, specifically where this filter constrains the London_Borough, as an argument. For example, one admissible argument should be: dfg[dfg['London_Borough']=='Camden'].
2. Get the Average Price for that Borough, for the years 1998 and 2018.
4. Calculate the ratio of the Average Price for 1998 divided by the Average Price for 2018.
5. Return that ratio.

Once you've written this function, you ultimately want to use it to iterate through all the unique London_Boroughs and work out the ratio capturing the difference of house prices between 1998 and 2018.

Bear in mind: you don't have to write a function like this if you don't want to. If you can solve the brief otherwise, then great! 

***Hint***: This section should test the skills you acquired in:
- Python Data Science Toolbox - Part One, all modules

```python
#creating function to calculate avg price ratio (year 2018 / year 1998) for any london borough from the dataframe
def create_price_ratio(lb):
    ap1998 = float(lb['Avg_Price'][lb['Year']==1998])
    ap2018 = float(lb['Avg_Price'][lb['Year']==2018])
    pr = [ap2018/ap1998]
    return pr
```

```python
#checking function output for a sample borough
create_price_ratio(dfg[dfg['London_Borough']=='Camden'])
```

```python
#creating empty dictionary to assign boroughs and their avg price ratio as valuese
lb_ap={}
```

```python
#creating dictionary entries by looping through all london broughs and their price ratios using create price ratio function and grouped dataframe
for b in dfg['London_Borough'].unique():
    ub=dfg[dfg['London_Borough']==b]
    lb_ap[b]=create_price_ratio(ub)
print(lb_ap)
```

```python
#converting dictionary to dataframe for visualizations and further data analysis to derive conclusions
lb_apr = pd.DataFrame(lb_ap)
```

```python
#looking at created dataframe
lb_apr.head()
```

```python
#transposing dataframe to reflect Borough and Avg Price Ratio as column headers
lb_apr_t=lb_apr.T
```

```python
#resetting index to get a numeric row index
lb_price_ratios = lb_apr_t.reset_index()
```

```python
#inspecting transposed dataframe
lb_price_ratios.head()
```

```python
#renaming column headers to Borough and 2018 (for price ratio of year 2018 divided by 1998)
lb_price_ratios.rename(columns = {'index' : 'Borough', 0 : '2018'}, inplace=True)
```

```python
#inspecting dataframe first few observations and column headers
lb_price_ratios.head()
```

```python
#creating sorted dataframe by high to low avg housing price ratio
lb_price_ratios_sorted_high_low = lb_price_ratios.sort_values(by=['2018'], ascending=False)
```

```python
#printing to check observations and draw conclusions from the sorted high/low price ratio dataframe
print(lb_price_ratios_sorted_high_low)
```

```python
#visualizing sorted high low price ratio dataframe for conclusions
ax=lb_price_ratios_sorted_high_low[['Borough','2018']].plot(kind='bar')
ax.set_xticklabels(lb_price_ratios_sorted_high_low.Borough)
```

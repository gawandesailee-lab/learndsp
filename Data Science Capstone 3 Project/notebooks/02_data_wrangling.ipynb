## 2.2 Introduction<a id='2.2_Introduction'></a>

### 2.2.1 Recap Of Data Science Problem<a id='2.2.1_Recap_Of_Data_Science_Problem'></a>

Electric Vehicle adoption has been growing significantly in the USA for the past decade. However, adoption is very uneven across different regions. Some have seen rapid adoption rates while others lag behind despite being in the same state.

The variation suggests that multiple factors such as socioeconomic, technology and policy incentives might be at play here and this project intends to explore their influence and how these factors interact with each other to drive adoption rates at region level.

Our objective is to develop a data-driven framework to quantify EV adoption at the county level and identify the key drivers influencing it. Using Washington state as a case study, the project constructs adoption rate as the proportion of electric vehicle registration to total passenger vehicle registration at the county year granularity.

The ultimate goal of the project is to provide actionable insights to help prioritize regions for EV infrastructure such as charging stations and incentives.

### 2.2.2 Datasets Used<a id='2.2.2_Datasets_Used'></a>

In this data wrangling step, data from the following key data sources for Washington State has been used.

1) Electric Vehicles Data : https://data.wa.gov/Transportation/Electric-Vehicle-Population-Data/f6w7-q2d2/about_data</p>
2) All Vehicle Registrations : https://data.wa.gov/Transportation/Vehicle-Registrations-by-Class-and-County/hmzg-s6q4/about_data </p>
3) Median Income : https://ofm.wa.gov/data-research/economy/median-household-income-estimates/</p>
4) Population : https://ofm.wa.gov/data-research/population-demographics/estimates/april-1-official/, https://ofm.wa.gov/data-research/population-demographics/estimates/archive/</p>
5) EV Charging Stations Data : https://ev-map-wsdot.hub.arcgis.com/apps/7e310dcd476640ec8c611c101f610c09/explore</p>

The median income, population & electric charging stations data has already been premerged into one csv file.

This step will involve cleansing, restructuring, reformating and narrowing down of datasets as needed to get a consistent and reliable dataset. The datasets will then be merged into a single dataset and validation checks will be done to make sure the final data is ready for Exploratory Data Analysis phase.

### 2.6 Merge All Datasets<a id='2.6_Merge_All_Datasets'></a>

So, we will need to filter out 2026 data from both datasets and also filter out data prior to 2017 from ev_data.

Next, we will go ahead and aggregate the EV Dataset at county year level to get Annual EV Count, Avg. Electric Range & CAFV Eligibility Rate.

```python
ev_data['bev_range_updated'] = np.where(ev_data['electric_vehicle_type'].str.contains('Battery Electric', case=False, na=False),
    ev_data['electric_range'],
    np.nan)
ev_data['phev_range_updated'] = np.where(ev_data['electric_vehicle_type'].str.contains('Plug-in Hybrid', case=False, na=False),
    ev_data['electric_range'],
    np.nan)

ev_agg= ev_data.groupby(['county','model_year']).agg(
    bev_ratio_within_ev = ('electric_vehicle_type', lambda x : x.str.contains('Battery', case=False, na=False).mean()),
    range_missing_rate = ('electric_range',lambda x : (x==0).mean()),
    cafv_eligibility_rate = ('cafv_true_eligible', 'mean'),
    ev_count_by_model_year = ('dol_vehicle_id','count'),
    bev_avg_range=('bev_range_updated', 'mean'),
    phev_avg_range=('phev_range_updated', 'mean')
    ).reset_index()
```

```python
merged_df.info()
```

Next, we will do feature engineering and create some features and our target variable which will be our ev_adoption_rate.

```python
#EV Adoption Rate -- Target Variable (multiplication by 100 is to get percentage values)
merged_df['ev_adoption_rate']=100*merged_df['registered_electric_vehicles']/merged_df['total_registered_vehicles']

#EV Growth Rate -- Trend Metric (multiplication by 100 is to get percentage values)
merged_df=merged_df.sort_values(['county','year'])
merged_df['ev_growth_rate']=100*merged_df.groupby('county')['registered_electric_vehicles'].transform(lambda x: (x / (x.shift(1) + 1)) - 1)
#Replace any null values with 0
merged_df['ev_growth_rate']=merged_df['ev_growth_rate'].fillna(0)

#Has Growth? Create a binary flag 0 or 1 to indicate if there was growth
merged_df['has_growth']=(merged_df['ev_growth_rate']>0).astype(int)

#New EV Purchase Ratio (multiplication by 100 is to get percentage values)
merged_df['new_ev_purchase_ratio'] = 100*merged_df['ev_count_by_model_year'] / merged_df['registered_electric_vehicles']
merged_df['new_ev_purchase_ratio'] = merged_df['new_ev_purchase_ratio'].replace([np.inf, -np.inf], np.nan).fillna(0)

#EV Adoption Lag -> Will help identify EV Adoption momentum
merged_df=merged_df.sort_values(['county','year'])
merged_df['ev_adoption_lag']=merged_df.groupby('county')['ev_adoption_rate'].shift(1)

#Handle Nulls in EV Adoption lag for first year 2017 as an indicator for modeling downstream
merged_df['lag_missing_flag'] = merged_df['ev_adoption_lag'].isna().astype(int)

#Fill with mean(Ensures linear models do not get distorted)
#mean_value = merged_df['ev_adoption_lag'].mean()
merged_df['ev_adoption_lag'] = merged_df['ev_adoption_lag'].fillna(0)

#High Adoption Indicator -- Target Variable
merged_df['high_adoption']=(merged_df['ev_adoption_rate']>merged_df['ev_adoption_rate'].quantile(0.75)).astype('int')

#Adoption Segment -- for Business Insights
merged_df['adoption_segment']=pd.qcut(merged_df['ev_adoption_rate'], q=3, labels=['low','medium','high'])

#Median Income Per Vehicle
merged_df['income_per_vehicle']=merged_df['median_income']/merged_df['total_registered_vehicles']

#EV Charging Density (multiplication by 100 is to get percentage values)
merged_df['ev_charging_density']=100*merged_df['ev_charging_stations']/merged_df['population']

#EV Charging Density Flag for 0 values
merged_df['has_charging_infrastructure']=(merged_df['ev_charging_stations']>0).astype(int)

#EV Charging Stations Growth
#merged_df['charging_growth']=merged_df.groupby('county')['ev_charging_stations'].transform(lambda x: (x / (x.shift(1) + 1)) - 1)*100.fillna(0)
```

Note: EV adoption lag feature was constructed using strictly prior-year values to prevent temporal leakage.

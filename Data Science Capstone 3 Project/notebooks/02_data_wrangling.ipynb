## 2.2 Introduction<a id='2.2_Introduction'></a>

### 2.2.1 Recap Of Data Science Problem<a id='2.2.1_Recap_Of_Data_Science_Problem'></a>

Electric Vehicle adoption has been growing significantly in the USA for the past decade. However, adoption is very uneven across different regions. Some have seen rapid adoption rates while others lag behind despite being in the same state.

The variation suggests that multiple factors such as socioeconomic, technology and policy incentives might be at play here and this project intends to explore their influence and how these factors interact with each other to drive adoption rates at region level.

Our objective is to develop a data-driven framework to quantify EV adoption at the county level and identify the key drivers influencing it. Using Washington state as a case study, the project constructs adoption rate as the proportion of electric vehicle registration to total passenger vehicle registration at the county year granularity.

The ultimate goal of the project is to provide actionable insights to help prioritize regions for EV infrastructure such as charging stations and incentives.

### 2.2.2 Datasets Used<a id='2.2.2_Datsets_Used'></a>

In this data wrangling step, data from the following key data sources for Washington State has been used.

1) Electric Vehicles Data : https://data.wa.gov/Transportation/Electric-Vehicle-Population-Data/f6w7-q2d2/about_data<p>
2) All Vehicle Registrations : https://data.wa.gov/Transportation/Vehicle-Registrations-by-Class-and-County/hmzg-s6q4/about_data <p>
3) Median Income : https://ofm.wa.gov/data-research/economy/median-household-income-estimates/<p>
4) Population : https://ofm.wa.gov/data-research/population-demographics/estimates/april-1-official/, https://ofm.wa.gov/data-research/population-demographics/estimates/archive/<p>
5) EV Charging Stations Data : https://ev-map-wsdot.hub.arcgis.com/apps/7e310dcd476640ec8c611c101f610c09/explore<p>

The median income, population & electric charging stations data has already been premerged into one csv file.

This step will involve cleansing, restructuring, reformating and narrowing down of datasets as needed to get a consistent and reliable dataset. The datasets will then be merged into a single dataset and validation checks will be done to make sure the final data is ready for Exploratory Data Analysis phase.

### 2.6 Merge All Datasets<a id='2.6_Merge_All_Datasets'></a>

First, we need to make sure both datasets have the same time period. reg_total has data from January 2017 to April 2026 while ev_data has data ranging from 1999 to 2027. So, we will need to filter out 2026 data from both datasets and also filter out data prior to 2017 from ev_data.

```python
reg_total=reg_total[reg_total['transaction_year']!=2026]
```

```python
ev_data=ev_data[(ev_data['model_year']>=2017) & (ev_data['model_year']<2026)]
```

```python
reg_total.describe()
```

```python
ev_data.describe(include='all')
```

Next, we will go ahead and aggregate the EV Dataset at county year level to get Annual EV Count, Avg. Electric Range & CAFV Eligibility Rate.

```python
ev_agg_bev_range = ev_data[(ev_data['electric_range']!=0) & (ev_data['electric_vehicle_type'].str.contains('Battery Electric',case=False, na=False))].groupby(['county','model_year']).agg(
    bev_avg_range = ('electric_range','mean')
).reset_index()

ev_agg_phev_range = ev_data[(ev_data['electric_range']!=0) & (ev_data['electric_vehicle_type'].str.contains('Plug-in Hybrid Electric',case=False, na=False))].groupby(['county','model_year']).agg(
    phev_avg_range = ('electric_range','mean')
).reset_index()

ev_bev_share = ev_data.groupby(['county','model_year']).agg(
    bev_share= ('electric_vehicle_type', lambda x : x.str.contains('Battery', case=False, na=False).mean())
).reset_index()

range_quality = (ev_data.groupby(['county','model_year']).agg(
    range_missing_rate = ('electric_range',lambda x : (x==0).mean())).reset_index())

ev_agg_other = ev_data.groupby(['county','model_year']).agg({
    'dol_vehicle_id': 'count',
    'cafv_binary': 'mean'
}).reset_index()

ev_agg_w_bev = pd.merge(ev_agg_other, ev_agg_bev_range, on = ['county','model_year'], how='left')

ev_agg_w_bev_phev = pd.merge(ev_agg_w_bev, ev_agg_phev_range, on = ['county','model_year'], how='left')

ev_agg_w_bev_share = pd.merge(ev_agg_w_bev_phev, ev_bev_share, on = ['county','model_year'], how='left')

ev_agg = pd.merge(ev_agg_w_bev_share, range_quality, on = ['county','model_year'], how='left')

ev_agg.rename(columns={'dol_vehicle_id':'ev_count_by_model_year',
                               'cafv_binary':'cafv_eligibility_rate'}, inplace=True)
```

```python
ev_agg.county.unique()
```

Finally, let's merge the 2 datasets.

```python
#merging ev_agg & reg_total datasets
merged_df1 = reg_total.merge(ev_agg, left_on=['residential_county','transaction_year'], right_on=['county','model_year'], how='left')
```

```python
merged_df1.info()
```

```python
#Lets replace null ev dataset columns values
merged_df1[['ev_count_by_model_year']]=merged_df1[['ev_count_by_model_year']].fillna(0)
merged_df1.loc[merged_df1['ev_count_by_model_year']==0, ['cafv_eligibility_rate','bev_share','range_missing_rate']]=0
```

```python
merged_df1.info()
```

Let's rename model_year to year and drop the repeated fields transaction_year and residential_county.

```python
merged_df1=merged_df1.drop(columns=['model_year','county'])
merged_df1=merged_df1.rename(columns={'transaction_year':'year', 'residential_county':'county'})
```

```python
merged_df1.info()
```

Next we will merge the supporting dataset containing population, median_income and ev charging stations count by county and year. This preprocessed support dataset has already been filtered to contain only years from 2017-2025.

```python
merged_df = merged_df1.merge(support_data, on=['county','year'], how='left')
```

```python
merged_df.info()
```

```python
merged_df.isnull().sum()
```

Next, we will do feature engineering and create some features and our target variable which will be our ev_adoption_rate.

```python
#EV Adoption Rate -- Target Variable
merged_df['ev_adoption_rate']=merged_df['registered_electric_vehicles']/merged_df['total_registered_vehicles']

#EV Growth Rate -- Trend Metric
merged_df=merged_df.sort_values(['county','year'])
merged_df['ev_growth_rate']=merged_df.groupby('county')['registered_electric_vehicles'].pct_change()
#Fill out any nulls in EV Growth Rate
merged_df['ev_growth_rate']=merged_df['ev_growth_rate'].fillna(0)

#High Adoption Indicator -- Target Variable
merged_df['high_adoption']=(merged_df['ev_adoption_rate']>merged_df['ev_adoption_rate'].quantile(0.75)).astype('int')

#Adoption Segment -- for Business Insights
merged_df['adoption_segment']=pd.qcut(merged_df['ev_adoption_rate'], q=3, labels=['low','medium','high'])

#Median Income Per Vehicle
merged_df['income_per_vehicle']=merged_df['median_income']/merged_df['total_registered_vehicles']
```
